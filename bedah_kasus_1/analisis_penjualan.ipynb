```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

np.random.seed(42)

# Katalog produk. Kalau tugas harus persis 2 produk (Laptop & Buku),
# hapus baris selain Laptop dan Buku.
produk_list = [
    {"Produk": "Laptop",     "Kategori": "Elektronik",  "Harga_Dasar": 8500000, "Qty_Min": 1,  "Qty_Max": 20},
    {"Produk": "Buku",       "Kategori": "Pendidikan",  "Harga_Dasar": 75000,   "Qty_Min": 50, "Qty_Max": 200},
    {"Produk": "Smartphone", "Kategori": "Elektronik",  "Harga_Dasar": 5000000, "Qty_Min": 1,  "Qty_Max": 30},
    {"Produk": "Tablet",     "Kategori": "Elektronik",  "Harga_Dasar": 3500000, "Qty_Min": 1,  "Qty_Max": 25},
    {"Produk": "Pulpen",     "Kategori": "Pendidikan",  "Harga_Dasar": 5000,    "Qty_Min": 100,"Qty_Max": 500},
    {"Produk": "Tas",        "Kategori": "Pendidikan",  "Harga_Dasar": 150000,  "Qty_Min": 10, "Qty_Max": 80},
    {"Produk": "Monitor",    "Kategori": "Elektronik",  "Harga_Dasar": 2000000, "Qty_Min": 1,  "Qty_Max": 40},
    {"Produk": "Kertas A4",  "Kategori": "Pendidikan",  "Harga_Dasar": 50000,   "Qty_Min": 20, "Qty_Max": 150},
]

n = 1000
rows = []
start_date = datetime(2024, 1, 1)
end_date = datetime(2024, 12, 31)

for _ in range(n):
    p = np.random.choice(produk_list)
    # Harga bervariasi ±10% dari harga dasar
    harga = int(p["Harga_Dasar"] * np.random.uniform(0.9, 1.1))
    qty = np.random.randint(p["Qty_Min"], p["Qty_Max"] + 1)
    # Tanggal acak sepanjang 2024
    delta = end_date - start_date
    random_days = np.random.randint(0, delta.days + 1)
    tanggal = start_date + timedelta(days=random_days)
    rows.append({
        "Produk": p["Produk"],
        "Kategori": p["Kategori"],
        "Harga": harga,
        "Jumlah_Terjual": qty,
        "Tanggal": tanggal
    })

df = pd.DataFrame(rows)
df["Total_Penjualan"] = df["Harga"] * df["Jumlah_Terjual"]
df.to_csv("data_penjualan_sintesis.csv", index=False)

print("Data berhasil dibuat:", df.shape)
df.head()
```
