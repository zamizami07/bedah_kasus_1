```python
import pandas as pd
import numpy as np

np.random.seed(42)

# ================= KATALOG JURUSAN =================
jurusan_list = [
    {"Jurusan": "Sistem Informasi",   "Fakultas": "Ilmu Komputer", "IPK_Mean": 3.45, "IPK_Std": 0.30, "Pria_Ratio": 0.55},
    {"Jurusan": "Teknik Informatika", "Fakultas": "Ilmu Komputer", "IPK_Mean": 3.40, "IPK_Std": 0.35, "Pria_Ratio": 0.70},
    {"Jurusan": "Manajemen",          "Fakultas": "Ekonomi",       "IPK_Mean": 3.50, "IPK_Std": 0.28, "Pria_Ratio": 0.45},
    {"Jurusan": "Akuntansi",          "Fakultas": "Ekonomi",       "IPK_Mean": 3.55, "IPK_Std": 0.25, "Pria_Ratio": 0.40},
    {"Jurusan": "Psikologi",          "Fakultas": "Psikologi",     "IPK_Mean": 3.60, "IPK_Std": 0.27, "Pria_Ratio": 0.30},
    {"Jurusan": "Hukum",              "Fakultas": "Hukum",         "IPK_Mean": 3.30, "IPK_Std": 0.35, "Pria_Ratio": 0.50},
    {"Jurusan": "Kedokteran",         "Fakultas": "Kedokteran",    "IPK_Mean": 3.50, "IPK_Std": 0.30, "Pria_Ratio": 0.45},
    {"Jurusan": "Teknik Sipil",       "Fakultas": "Teknik",        "IPK_Mean": 3.20, "IPK_Std": 0.40, "Pria_Ratio": 0.75},
]

# ================= GENERATE DATA =================
n = 1000
rows = []

for _ in range(n):
    j = np.random.choice(jurusan_list)
    
    semester = np.random.randint(1, 9)
    
    # Usia berkorelasi dengan semester
    usia_min = 17 + (semester - 1) // 2
    usia = np.random.randint(usia_min, usia_min + 3)
    
    # IPK: distribusi normal dibatasi 2.0 - 4.0
    ipk = np.random.normal(j["IPK_Mean"], j["IPK_Std"])
    ipk = np.clip(ipk, 2.0, 4.0)
    ipk = round(ipk, 2)
    
    gender = "Laki-laki" if np.random.random() < j["Pria_Ratio"] else "Perempuan"
    
    rows.append({
        "Jurusan": j["Jurusan"],
        "Fakultas": j["Fakultas"],
        "Semester": semester,
        "IPK": ipk,
        "Usia": usia,
        "Gender": gender
    })

df = pd.DataFrame(rows)
df = df[["Jurusan", "Fakultas", "Semester", "IPK", "Usia", "Gender"]]

df.to_csv("data_mahasiswa_sintesis.csv", index=False)

print("Data berhasil dibuat:", df.shape)
print("\nPreview:")
print(df.head(10))
print("\nStatistik:")
print(df.describe())
```
