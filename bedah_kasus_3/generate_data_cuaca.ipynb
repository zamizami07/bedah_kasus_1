```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

np.random.seed(42)

kota_list = [
    {"Kota": "Jakarta",   "Suhu_Mean": 32.0, "Suhu_Std": 2.0, "Lembab_Mean": 75, "Hujan_Prob": 0.45},
    {"Kota": "Bandung",   "Suhu_Mean": 24.0, "Suhu_Std": 3.0, "Lembab_Mean": 80, "Hujan_Prob": 0.50},
    {"Kota": "Surabaya",  "Suhu_Mean": 33.0, "Suhu_Std": 1.5, "Lembab_Mean": 70, "Hujan_Prob": 0.30},
    {"Kota": "Medan",     "Suhu_Mean": 31.0, "Suhu_Std": 2.0, "Lembab_Mean": 82, "Hujan_Prob": 0.55},
    {"Kota": "Makassar",  "Suhu_Mean": 30.5, "Suhu_Std": 1.8, "Lembab_Mean": 78, "Hujan_Prob": 0.40},
    {"Kota": "Yogyakarta","Suhu_Mean": 28.5, "Suhu_Std": 2.5, "Lembab_Mean": 76, "Hujan_Prob": 0.48},
    {"Kota": "Denpasar",  "Suhu_Mean": 29.5, "Suhu_Std": 1.5, "Lembab_Mean": 74, "Hujan_Prob": 0.35},
    {"Kota": "Balikpapan","Suhu_Mean": 30.0, "Suhu_Std": 1.5, "Lembab_Mean": 85, "Hujan_Prob": 0.60},
]

def pilih_kondisi(hujan_prob):
    r = np.random.random()
    if r < hujan_prob:
        return "Hujan"
    elif r < hujan_prob + 0.35:
        return "Berawan"
    else:
        return "Cerah"

n = 1000
rows = []
start_date = datetime(2024, 1, 1)
end_date = datetime(2024, 12, 31)
delta = end_date - start_date

for _ in range(n):
    k = np.random.choice(kota_list)
    
    suhu = np.random.normal(k["Suhu_Mean"], k["Suhu_Std"])
    suhu = round(suhu, 1)
    
    kelembaban = np.random.normal(k["Lembab_Mean"], 10)
    kelembaban = np.clip(kelembaban, 20, 100)
    kelembaban = int(round(kelembaban))
    
    random_days = np.random.randint(0, delta.days + 1)
    tanggal = start_date + timedelta(days=random_days)
    
    kondisi = pilih_kondisi(k["Hujan_Prob"])
    
    if kondisi == "Hujan":
        kelembaban = min(100, kelembaban + np.random.randint(5, 15))
        suhu = round(suhu - np.random.uniform(1, 3), 1)
    elif kondisi == "Cerah":
        kelembaban = max(20, kelembaban - np.random.randint(5, 15))
        suhu = round(suhu + np.random.uniform(0.5, 2), 1)
    
    rows.append({
        "Kota": k["Kota"],
        "Tanggal": tanggal,
        "Suhu_C": suhu,
        "Kelembaban_Persen": kelembaban,
        "Kondisi": kondisi
    })

df = pd.DataFrame(rows)
df = df.sort_values("Tanggal").reset_index(drop=True)

df.to_csv("data_cuaca_sintesis.csv", index=False)

print("Data berhasil dibuat:", df.shape)
print("\nPreview:")
print(df.head(10))
print("\nDistribusi Kondisi:")
print(df["Kondisi"].value_counts())
```
